```python
import pandas as pd
df = pd.read_csv(r"C:\Users\DELL\Downloads\dataset_part_2.csv")
print(df.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart 1: Flight Number vs. Payload Mass
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Pay load Mass (kg)", fontsize=20)
plt.show()

# Task 1: Flight Number vs. Launch Site
sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.show()

# Task 2: Payload vs. Launch Site
sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Pay load Mass (kg)", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.show()

# Task 3: success rate of each orbit
success_rate = df.groupby('Orbit')['Class'].mean().reset_index()
plt.figure(figsize=(10, 5))
sns.barplot(x='Orbit', y='Class', data=success_rate)
plt.xlabel("Orbit", fontsize=14)
plt.ylabel("Success Rate", fontsize=14)
plt.show()
print(success_rate)

# Task 4: Flight Number vs. Orbit
sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.show()

# Task 5: Payload vs. Orbit
sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Pay load Mass (kg)", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.show()

# Task 6: yearly average success rate
df['Year'] = df['Date'].str.split('-').str[0]
yearly = df.groupby('Year')['Class'].mean().reset_index()
plt.figure(figsize=(10, 5))
sns.lineplot(x='Year', y='Class', data=yearly, marker='o')
plt.xlabel("Year", fontsize=14)
plt.ylabel("Average Success Rate", fontsize=14)
plt.show()
print(yearly)

# Task 7: select features
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins',
               'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]

# Task 8: one-hot encode
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])

# Task 9: float64 and export
features_one_hot = features_one_hot.astype('float64')
features_one_hot.to_csv('dataset_part_3.csv', index=False)
print(features_one_hot.shape)
```
