```python
import pandas as pd
import numpy as np

# Load the dataset collected in the previous lab
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
print(df.head(10))

# Percentage of missing values per column
print("\nMissing values (%):")
print(df.isnull().sum() / len(df) * 100)

# Data types
print("\nData types:")
print(df.dtypes)

# Task 1: launches per launch site
print("\nLaunches per site:")
print(df['LaunchSite'].value_counts())

# Task 2: launches per orbit type
print("\nLaunches per orbit:")
print(df['Orbit'].value_counts())

# Task 3: mission outcome counts
landing_outcomes = df['Outcome'].value_counts()
print("\nLanding outcomes:")
print(landing_outcomes)

for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)

# Bad outcomes: the landing did not succeed
bad_outcomes = {'False Ocean', 'False RTLS', 'False ASDS', 'None None', 'None ASDS'}
print("\nBad outcomes:", bad_outcomes)

# Task 4: landing class (0 = bad outcome, 1 = good outcome)
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class
print(df[['Class']].head(8))
print(df.head(5))

# Task 5: success rate and export
print("\nSuccess rate:", df["Class"].mean())
df.to_csv('dataset_part_2.csv', index=False)
```
